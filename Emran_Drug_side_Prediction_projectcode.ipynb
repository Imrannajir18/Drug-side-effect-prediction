```
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.neural_network import MLPClassifier
from sklearn import svm
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import VotingClassifier
```

```
from google.colab import drive
drive.mount('/content/drive')
```

```
dataset_path = "/content/drive/My Drive/Datasets.csv"
```

```
df = pd.read_csv(dataset_path, encoding='latin-1')
```

```
def apply_results(Rating):
    return 0 if int(Rating) <= 7 else 1  # 0: Low Side Effect, 1: High Side Effect

df['Results'] = df['Rating'].apply(apply_results)
```

```
cv = CountVectorizer()
X = df['UID'].apply(str)  # Convert UID to string
y = df['Results']
X = cv.fit_transform(X)
```

```
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```
models = []
```

```
mlpc = MLPClassifier().fit(X_train, y_train)
y_pred_mlp = mlpc.predict(X_test)
print("MLP Classifier Accuracy:", accuracy_score(y_test, y_pred_mlp) * 100)
models.append(('MLPClassifier', mlpc))
```

```
lin_clf = svm.LinearSVC()
lin_clf.fit(X_train, y_train)
y_pred_svm = lin_clf.predict(X_test)
print("SVM Accuracy:", accuracy_score(y_test, y_pred_svm) * 100)
models.append(('SVM', lin_clf))
```

```
reg = LogisticRegression(random_state=0, solver='lbfgs')
reg.fit(X_train, y_train)
y_pred_logistic = reg.predict(X_test)
print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_logistic) * 100)
models.append(('Logistic', reg))
```

```
classifier = VotingClassifier(models)
classifier.fit(X_train, y_train)
y_pred_voting = classifier.predict(X_test)
print("Voting Classifier Accuracy:", accuracy_score(y_test, y_pred_voting) * 100)
```

```
uid_input = input("Enter UID for Prediction: ")
vector_input = cv.transform([uid_input]).toarray()
predicted_effect = classifier.predict(vector_input)
```

```
if predicted_effect[0] == 0:
    print("Prediction: Low Side Effect Found")
else:
    print("Prediction: High Side Effect Found")
```
